```python

# ============================================
# PROJECT 02: NETFLIX DATA VISUALIZATION
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Custom Netflix-inspired theme
sns.set_theme(style="darkgrid")

# Custom color palette
netflix_colors = [
    "#E50914",
    "#B20710",
    "#6366F1",
    "#A855F7",
    "#06B6D4",
    "#F59E0B"
]

# Chart formatting
plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.facecolor"] = "#141414"
plt.rcParams["axes.facecolor"] = "#202020"
plt.rcParams["axes.labelcolor"] = "white"
plt.rcParams["axes.titlecolor"] = "#E50914"
plt.rcParams["xtick.color"] = "white"
plt.rcParams["ytick.color"] = "white"
plt.rcParams["text.color"] = "white"

print("Netflix Data Visualization Project Started!")
print("All libraries imported successfully!")
```

```python

# Load Netflix Dataset

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"

df = pd.read_csv(url)

# Display first 10 records
print("NETFLIX DATASET")
display(df.head(10))

# Dataset dimensions
print("\nDataset Shape:", df.shape)

# Column names
print("\nDataset Columns:")
print(df.columns.tolist())
```

```python

# Dataset Information

print("DATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe())

print("\nMISSING VALUES")
display(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nCONTENT TYPES")
print(df["type"].value_counts())
```

```python

# Create a working copy
netflix = df.copy()

# Remove duplicate records
netflix.drop_duplicates(inplace=True)

# Handle missing values
netflix["director"] = netflix["director"].fillna("Unknown")
netflix["cast"] = netflix["cast"].fillna("Unknown")
netflix["country"] = netflix["country"].fillna("Unknown")
netflix["rating"] = netflix["rating"].fillna("Not Rated")

# Convert date column
netflix["date_added"] = pd.to_datetime(
    netflix["date_added"].str.strip(),
    errors="coerce"
)

# Extract duration in minutes for movies
netflix["duration_min"] = pd.to_numeric(
    netflix["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)

print("Data Cleaning Completed Successfully!")
display(netflix.head())
```

```python

# STEP 5: MOVIES VS TV SHOWS

content_counts = netflix["type"].value_counts()

plt.figure(figsize=(9, 6))

bars = plt.bar(
    content_counts.index,
    content_counts.values,
    color=["#E50914", "#6366F1"],
    width=0.5
)

# Add values above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 50,
        f"{int(height)}",
        ha="center",
        fontsize=12,
        fontweight="bold",
        color="white"
    )

plt.title("Netflix Movies vs TV Shows", fontsize=17, fontweight="bold")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# STEP 6: RELEASE YEAR DISTRIBUTION

plt.figure(figsize=(11, 6))

sns.histplot(
    data=netflix,
    x="release_year",
    bins=30,
    color="#E50914",
    edgecolor="white"
)

plt.title("Netflix Content Release Year Distribution",
          fontsize=17, fontweight="bold")

plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()
```

```python

# STEP 7: TOP 10 COUNTRIES

country_data = (
    netflix[netflix["country"] != "Unknown"]["country"]
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(11, 6))

sns.barplot(
    x=country_data.values,
    y=country_data.index,
    hue=country_data.index,
    palette="Reds_r",
    legend=False
)

plt.title("Top 10 Countries Producing Netflix Content",
          fontsize=17, fontweight="bold")

plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python

# STEP 8: MOVIE DURATION BOX PLOT

movie_data = netflix[
    (netflix["type"] == "Movie") &
    (netflix["duration_min"].notna())
]

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=movie_data,
    y="duration_min",
    color="#E50914",
    width=0.4
)

plt.title("Netflix Movie Duration Analysis",
          fontsize=17, fontweight="bold")

plt.ylabel("Duration (Minutes)")
plt.xlabel("Netflix Movies")

plt.tight_layout()
plt.show()
```

```python

# STEP 9: GENRE DISTRIBUTION

genres = (
    netflix["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=genres.values,
    y=genres.index,
    hue=genres.index,
    palette="Reds_r",
    legend=False
)

plt.title("Top 10 Netflix Genres",
          fontsize=17, fontweight="bold")

plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.tight_layout()
plt.show()
```

```python

# STEP 10: CORRELATION HEATMAP

correlation_data = netflix[
    ["release_year", "duration_min"]
].dropna()

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="RdBu_r",
    fmt=".2f",
    linewidths=1,
    square=True
)

plt.title("Netflix Content Correlation Heatmap",
          fontsize=16, fontweight="bold")

plt.tight_layout()
plt.show()
```

```python

# STEP 11: SCATTER PLOT

scatter_data = netflix[
    (netflix["type"] == "Movie") &
    (netflix["duration_min"].notna())
]

plt.figure(figsize=(11, 6))

sns.scatterplot(
    data=scatter_data,
    x="release_year",
    y="duration_min",
    color="#E50914",
    alpha=0.6,
    s=45
)

plt.title(
    "Netflix Movie Release Year vs Duration",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("Release Year")
plt.ylabel("Movie Duration (Minutes)")

plt.tight_layout()
plt.show()
```
